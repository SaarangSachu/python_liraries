Series = a pandas 1 dimensional labeled array that cab hold any data type think of it like a single column in a spreadsheet (1=Dimensional)

```python
data=[100,102,104]
series=pd.Series(data) #Series is a consrtuctor
print(series)
```

```python
data=[100,102,104]
series=pd.Series(data, index=["a","b","c"]) #Series is a consrtuctor
print(series)
```

```python
data=[100,102,104]
series=pd.Series(data, index=["a","b","c"]) #Series is a consrtuctor
print(series.loc["a"])
print(series.loc["b"])
print(series.loc["c"])

series.loc['c']=200 #updating the value
print(series.loc["c"])

print(series.iloc[0])# iloc is for integer location

```

```python
data=[100,102,104]
series=pd.Series(data, index=["a","b","c"]) #Series is a consrtuctor

print(series[series>=101])
print(series[series<=200])

```

```python
calories={"day1":1750, 'day2':2100,'day3':1700}

series=pd.Series(calories)

print(series)

print(series.loc["day1"])
series.loc['day3']+=500
print(series.loc['day3'])
print()
print(series)
```

```python
calories={"day1":1750, 'day2':2100,'day3':1700}

series=pd.Series(calories)

print(series[series>=2000])
print()
print(series[series<=2000])
```

```python
data={
    "name":['bob','patrik','tom'],
    "age":[20,30,40]
    }


df = pd.DataFrame(data,index=['emp1','emp2','emp3'])#constructor

df["job"]=['manager','driver','cook']


new_row=pd.DataFrame([{'name':'jon','age':23,'job':"emgineer"}],
                        index=['emp4'])
df=pd.concat([df,new_row])
print(df)
```

```python
# add new rows
new_rows=pd.DataFrame([{'name':'huku','age':20,'job':"baker"},
                        {'name':'crab','age':60,'job':"manager"},
                        {'name':'huku','age':20,'job':"baker"}],
                        index=['emp5','emp6','emp7'])

df=pd.concat([df,new_rows])
print(df)
```
